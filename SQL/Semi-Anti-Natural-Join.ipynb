## Création des données

```python
users = {
    'User_ID': [123, 124, 125],
    'User_Name': ['Noah', 'Elio', 'Manon'],
}

users = pd.DataFrame(users)
users
```

```python

orders = {
    'User_ID': [123, 125, 126],
    'Order_ID': [333, 222, 111],
}

orders = pd.DataFrame(orders)
orders
```

## Semi Join

```python
query = """
select *
from users U
SEMI JOIN orders O
ON (U.User_ID = O.User_ID);
"""

duckdb.sql(query).df()
```

## Anti Join

```python
query = """
select *
from users U
ANTI JOIN orders O
ON (U.User_ID = O.User_ID);
"""

duckdb.sql(query).df()
```

## Natural Join

```python
query = """
select *
from users U
NATURAL JOIN orders O;
"""

duckdb.sql(query).df()
```
